Measure of central Tendency

```python
import numpy as np
from scipy import stats
x=[100,110,105,120,115,110,130,125,110,175]
print("Mean:",np.mean(x))
print("Median:",np.median(x))
print("Mode:",stats.mode(x,keepdims=False))
```

Measure of Dispersion

```python
print("Range:",np.ptp(x))
print("Q1:",np.percentile(x,25))
print("Q2:",np.percentile(x,50))
print("Q3:",np.percentile(x,75))
print("IQR:",np.percentile(x,75)-np.percentile(x,25))
print("Variance:",np.var(x,ddof=1))
print("Standard Deviation :",np.std(x,ddof=1))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.boxplot(x=x)
plt.title("Production Distribution")
plt.show()
```

Symmetry and Skewness

```python
from scipy.stats import skew
print("Skewness :",skew(x))
sns.histplot(x,kde=True)
plt.show()
```

Final Project Complete Python Program

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import skew

# STEP 1: Create Dataset
df = pd.DataFrame({
    "Day": range(1, 11),
    "Production": [100,110,105,120,115,110,130,125,110,175],
    "Rejection": [5,6,4,8,7,5,9,8,6,20],
    "Downtime": [30,25,35,20,22,28,18,20,26,60]
})

# STEP 2: Basic Information
print(df.head())
print(df.info())
print(df.describe())

# STEP 3: Central Tendency
for col in ["Production", "Rejection", "Downtime"]:
    print("\n", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode().tolist())

# STEP 4: Dispersion
for col in ["Production", "Rejection", "Downtime"]:
    print("\n", col)
    print("Range:", df[col].max() - df[col].min())
    print("Variance:", df[col].var())
    print("Standard Deviation:", df[col].std())
    print("IQR:", df[col].quantile(.75)
          - df[col].quantile(.25))

# STEP 5: Percentile
print("\n90th Percentile:")
print(df["Production"].quantile(.90))

# STEP 6: Skewness
print("\nSkewness:")
print(df.skew(numeric_only=True))

# STEP 7: Covariance & Correlation
print("\nCovariance:")
print(df.cov(numeric_only=True))

print("\nCorrelation:")
print(df.corr(numeric_only=True))

# STEP 8: Visualization

# Histogram
df["Production"].hist()
plt.title("Production Distribution")
plt.show()

# Boxplot
sns.boxplot(x=df["Production"])
plt.title("Production Outliers")
plt.show()

# Heatmap
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()

# Pairplot
sns.pairplot(df.drop(columns="Day"))
plt.show()

# STEP 9: Rejection Rate
rejection_rate = (
    df["Rejection"].sum() /
    df["Production"].sum()
) * 100

print("Overall Rejection Rate:", rejection_rate, "%")
```
